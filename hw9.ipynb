Problem 1d

```python
import numpy as np

def simulate_once(L, alpha=1, beta=1):
    i = 0
    t = 0.0
    
    while i < L:
        if i == 0:
            rate = alpha
            t += np.random.exponential(1/rate)
            i += 1
        else:
            rate = alpha + beta
            t += np.random.exponential(1/rate)
            if np.random.rand() < alpha / rate:
                i += 1
            else:
                i -= 1
                
    return t

def simulate_many(N, L):
    samples = np.array([simulate_once(L) for _ in range(N)])
    return samples.mean(), samples.var(), samples

N = 1000
L = 20

mean_est, var_est, samples = simulate_many(N, L)

print("Estimated mean:", mean_est)
print("Theoretical mean:", L*(L+1)/2)
print("Estimated variance:", var_est)
```

The simulated and theoretical mean are reasonably similar.

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def rand(m, b):
    m = int(np.floor(m))
    return 1/b * np.sum(1/np.arange(1, m))

def det(m, b):
    return 1/b * np.log(m)

m_vals = np.linspace(1, 1000)

r = np.array([rand(m, 1) for m in m_vals])
d = det(m_vals, 1)

plt.plot(m_vals, r, label="Yule process")
plt.plot(m_vals, d, label="Deterministic")
plt.xlabel("m")
plt.ylabel(r'$\tau_m$')
plt.legend()
plt.show()
```

It seems that asymptotically they grow at the same rate but differ by a constant.
