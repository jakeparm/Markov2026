```python
import numpy as np
q = 0.4
p = 0.35
s = 0.25

stats = []
for i in range(100000):
    money = 10
    while(money != 0):
        u = np.random.random()
        if u <=q:
            money -= 1
        elif u<= q+p:
            money += 1
        else:
            stats.append(money)
            break
        if money == 0:
            stats.append(money)
print(np.mean(stats))
```

Problem 2

Part (a)

```python

import math
P = np.zeros((6,6))
for i in range(6):
    for j in range(i+1):
        P[i,j] = math.comb(i,j)*(9/10)**j*(1/10)**(i-j)
P[0] = [0,0,0,0,0,1]
print(P)
```

Part (b)

```python
one = np.ones(5)
id = np.identity(5)
q = np.linalg.solve(id-P[1:,1:],one)
print(q[4])
```

Part (c)

```python
eigval, eigvec = np.linalg.eig(P.T)
index = np.isclose(eigval, 1)
pi = eigvec[:, index][:, 0].real
pi = pi / np.sum(pi)
print(pi)
```
